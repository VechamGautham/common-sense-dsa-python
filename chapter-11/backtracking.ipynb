  ## BackTracking 

  - back tracking is a systematic trail and error technique used to solve problems by exploring all possible choices , and undoing (backtracking) a choice when it leads to a dead end 
  - it builds a solution step by step, whenever a partial solution cannot be extended further, it reverts that step and tries another path 

  - eg imagine a robot navigating a maze trying to reach the exit , 
      1. the robot moves forward step by step 
      2. if it reaches a dead end , it backtracks goes back to the previous decision point 
      3. it choooses a different unexplored path 
      4. this continues until:
            - it finds the exit, or 
            - all paths are exhaused (meaning no solution exists)
