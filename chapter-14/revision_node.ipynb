## Reading

```python
class LinkedList:
    def __init__(self,first_node = None):
        self.first_node = first_node 
    
    def read(self,index):
        current_node = self.first_node 
        current_index = 0 

        while current_index < index :
            current_node = current_node.next_node 
            current_index += 1 

            if not current_node :
                return None 
        
        return current_node.data 


list = LinkedList(node_1)

print(list.first_node.data)

print(list.read(4))

```

## Doubly Linked List

```python
class Node:
    def __init__(self,data):
        self.data = data 
        self.next_node = None 
        self.previous_node = None 
    
    
```

```python
class DoublyLinkedList:
    def __init__(self,first_node=None,last_node=None):
        self.first_node = first_node
        self.last_node = last_node 
    
    def append(self,value):
        new_node = Node(value)

        if not self.first_node:
            self.first_node = new_node 
            self.last_node = new_node 
        
        else:
            new_node.previous_node = self.last_node 
            self.last_node.next_node = new_node 
            self.last_node = new_node 
        
    def pop_head(self):
        if self.first_node is None :
            return 
        
        popped_head = self.first_node 

        if self.first_node == self.last_node:
            self.first_node = None 
            self.last_node = None 
            return popped_head
        
        self.first_node = self.first_node.next_node 
        self.first_node.previous_node = None 
        return popped_head 
    
    def to_list(self):
        out = []
        curr = self.first_node
        while curr is not None:
            out.append(curr.data)
            curr = curr.next_node
        return out 
    
    

    

        
```

```python
class Queue:
    def __init__(self):
        self.data = DoublyLinkedList()
    
    def enqueue(self,element):
        self.data.append(element)
    
    def dequeue(self):
        dequeued_node = self.data.pop_head()
        return dequeued_node.data 
    
    def read(self):
        if not self.data.first_node:
            return None 
        return self.data.first_node.data 


dll = DoublyLinkedList()
dll.append(10)
dll.append(20)
dll.append(30)

print(dll.to_list())

```
