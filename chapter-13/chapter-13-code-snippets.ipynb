# Recursive Algorithams for Speed

## Quicksort
- Quicksort is an extremely fast sorting algorithm that is paticurly efficent for average scenarios. While in worst case scenarios (that is iversely sorted array) it performs similarly to insertion sort and selection sort , it's much faster for average scenarios - which are what occur most fo the time 

- Quicksort relies on a concept called partitioning
- The Quicksort algorithm is a combination of partitions and recursion.

## Partitioning

- To partition an array is to take a random value from the array which is then called pivot  and make sure that every number that is less than the pivot ends up to the left of the pivot and tha every number greater than the pivot ends up to the right of the pivot 



## Code Implementation : Partitioning

```python
class SortableArray:
    def __init__(self,array):
        self.array = array 
    
    def partition(self,left_pointer,right_pointer):
        pivot_index = right_pointer 
        pivot = self.array[pivot_index]
        right_pointer -= 1 

        while True :
            while  self.array[left_pointer] < pivot:
                left_pointer += 1 
            while self.array[right_pointer] > pivot :
                right_pointer -= 1 
            
            if left_pointer >= right_pointer :
                break 
            else:
                self.array[left_pointer],self.array[right_pointer] = self.array[right_pointer],self.array[left_pointer]
                left_pointer += 1
                right_pointer -= 1 
        
        self.array[left_pointer],self.array[pivot_index] =  self.array[pivot_index],self.array[left_pointer]

        return self.array 
        

array = [1,2,3,4,5,6]
obj = SortableArray(array=array)

obj.partition(left_pointer = 0,right_pointer = len(array) -1)

```
